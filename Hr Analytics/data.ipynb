```python

import  pandas as pd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Read Rhe Data Set
df = pd.read_csv("C:/Users/Prakash/Downloads/HR-Employee-Attrition.csv")
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
#data Cleaning
df["Attrition"] = df["Attrition"].map(
    {"Yes":1,"No":0}
)
```

```python
#Employee Age Distrubution

sns.histplot(df["Age"])

plt.show()
```

```python
#Monthly Income
sns.boxplot(x=df["MonthlyIncome"])

plt.show()
```

Attrition Analysis


```python
#total Attritions
df["Attrition"].value_counts()
```

```python
sns.countplot(x="Attrition",data=df)
plt.show()
```

```python
#Department Wise Attrition
pd.crosstab(
    df["Department"],
    df["Attrition"]
)
```

```python
sns.countplot(
    x="Department",
    hue="Attrition",
    data=df
)

plt.xticks(rotation=45)
plt.show()
```

```python
#Salary vs Attrition
sns.boxplot(
    x="Attrition",
    y="MonthlyIncome",
    data=df
)

plt.show()
```

```python
#overtime anlysis
pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)
```

```python
sns.countplot(
    x="OverTime",
    hue="Attrition",
    data=df
)

plt.show()
```

```python
#Job satasfaction Analysis
sns.boxplot(
    x="Attrition",
    y="JobSatisfaction",
    data=df
)

plt.show()
```

```python
#Correlation Analysis
plt.figure(figsize=(12,8))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    cmap='coolwarm',
    center=0,
    annot=False
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.show()
```
